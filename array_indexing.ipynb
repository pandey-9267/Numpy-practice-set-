```python
import numpy as np
```

```python
arr = np.arange(12, 37)
arr
```

```python
# NumPy Indexing and Slicing (1D Array)
```

```python
# INDEXING
# Access a single element using its index
print("value of this index:", arr[5])
```

```python
# SLICING
# Syntax: arr[start:end]
# Start index is included, but end index is excluded
print("Values from index 5 to 10:", arr[5:11])
# If the start index is omitted, slicing starts from index 0
print("Values from the beginning to index 8:", arr[:9])
# If the end index is omitted, slicing continues to the end
print("Values from index 17 to the end:", arr[17:])
# HOW SLICING WORKS WITH STEP
# Syntax: arr[start:end:step]
# start = starting index (included)
# end = ending index (excluded)
# step = interval between selected indices
print("Values from index 2 to 23 with a step of 2:", arr[2:24:2])
```

```python
# NumPy Indexing and Slicing of a Matrix
```

```python
# INDEXING
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
# Access the second row
print("Value of the second row:", arr[1])
# Access the third column
print("Values of the third column:", arr[:, 2])
# Access a single element at row 0, column 2
print("Value at row 0 and column 2:", arr[0, 2])
# Access a single element at row 4, column 4
print("Value at row 4 and column 4:", arr[4, 4])
```

```python
# SLICING OF A 2D ARRAY
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
# Select rows 0 to 2 and columns 1 to 2
print("Values from the first 3 rows and columns 2 to 3:\n", arr[0:3, 1:3])
# Select rows 0 to 2 and columns 3 to 4
print("Values from the first 3 rows and columns 4 to 5:\n", arr[3:, 3:])
```

```python
# BOOLEAN INDEXING
arr = np.arange(11, 21)
print("Original array:", arr)
# Create a Boolean mask for even numbers
bool_index = arr % 2 == 0
print("Boolean mask:", bool_index)
```

```python
# Select only the even numbers
even = arr[bool_index]
print("Even numbers from the array:", even)
```
