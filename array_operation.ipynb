```python
import numpy as np
```

```python
# ARITHMETIC OPERATIONS IN NUMPY
```

```python
# Create two NumPy arrays
a1 = np.array([1, 2, 3, 4, 5])
a2 = np.array([6, 7, 8, 9, 10])
```

```python
# Addition of two arrays
print("Addition of two arrays:\n", a1 + a2)
# Subtraction of two arrays
print("Subtraction of two arrays:\n", a1 - a2)
# Multiplication of two arrays (element-wise)
print("Multiplication of two arrays:\n", a1 * a2)
# Division of two arrays (element-wise)
print("Division of two arrays:\n", a1 / a2)
# Exponentiation (raise each element of a1 to the corresponding power in a2)
print("Exponentiation of two arrays:\n", a1 ** a2)
# Floor division of two arrays (discard the decimal part)
print("Floor division of two arrays:\n", a1 // a2)
```

```python
# BROADCASTING IN NUMPY
```

```python
# Create a NumPy array
arr = np.array([1, 2, 3, 4, 5, 6])
print("Original array:\n", arr)
```

```python
# Add 10 to each element using broadcasting
print("Array after adding 10 to each element:\n", arr + 10)
```

```python
# DEEP COPY AND SHALLOW COPY (VIEW) IN NUMPY
```

```python
# SHALLOW COPY (VIEW)

# Create the original array
arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)

# Create a view of the original array
view_arr = arr.view()
print("View of the original array:", view_arr)

# Modify the first element of the view
view_arr[0] = 100

# Display both arrays after modification
print("Original array after modifying the view:", arr)
print("View after modification:", view_arr)
```

```python
# DEEP COPY

# Create the original array
arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)

# Create an independent copy of the array
copy_arr = arr.copy()
print("Copied array:", copy_arr)

# Modify the first element of the copied array
copy_arr[0] = 100

# Display both arrays after modification
print("Original array after modifying the copy:", arr)
print("Copied array after modification:", copy_arr)
```
