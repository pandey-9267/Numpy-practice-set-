```python
# Array attributes
# In NumPy, array attributes are properties that give you information about an array, such as its dimensions, size, and data type
# always remember attributes never call by () this
arr = np.array([[1,2,3],[4,0,6],[7,8,9]])
arr
```

```python
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Total elements:", arr.size)
print("Data type:", arr.dtype)
print("Bytes per element:", arr.itemsize)
print("Total bytes:", arr.nbytes)
```

```python
# Array methods ?
# In NumPy, array methods are functions that perform operations on an array. Unlike attributes, methods usually perform an action or calculate something
arr = np.array([[1,2,3],[4,0,6],[7,8,9]])
arr
```

```python
print("Total sum:", arr.sum())
print("Min value:", arr.min())
print("Max value:", arr.max())
print("Sum of columns individually:", np.sum(arr, axis = 0))
print("Sum of rows individually:", np.sum(arr, axis = 1))
print("Index value of the last array:", arr.argmax())
print("Index value of the first array:", arr.argmin())
print("Mean:", arr.mean())
```

```python
# Low std() → numbers are closer to the mean.
# High std() → numbers are more spread out.
# std() = 0 → all numbers are identical.
arr.std()
```
